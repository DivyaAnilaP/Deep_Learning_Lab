```python
import torch
import torch.nn as nn
import torch.optim as optim
#torchvison-->all images will be there
from torchvision import datasets, transforms
from torch.utils.data import DataLoader
```

```python
transform=transforms.Compose([

    
    transforms.ToTensor(),
    transforms.Normalize((0.5,0.5,0.5),(0.5,0.5,0.5))#mean and std for i/p images --3channels(rgb)                        3chn3
])
```

## Download CIFAR-10 dataset


```python
train_data=datasets.CIFAR10(root='./data',train=True,download=True,transform=transform)
test_data=datasets.CIFAR10(root='./data',train=False,download=True,transform=transform)
```

## DataLoader 

```python
#dataloader takes data from dataset and feeds it to the n/w in batches
train_loader=DataLoader(train_data,batch_size=64,shuffle=True)#shuffle---to make sure n/w does not learn the order of data
test_loader=DataLoader(test_data,batch_size=64,shuffle=False)
```

## build the architecture

```python
import torch
import torch.nn as nn
import torch.nn.functional as F  # For max_pool2d

class cifar_cnn_pytorch(nn.Module):
    def __init__(self):
        super(cifar_cnn_pytorch, self).__init__()
        # More filters -> more features extracted
        # Filter size decreases as we go deeper
        self.conv1 = nn.Conv2d(in_channels=3, out_channels=6, kernel_size=5)
        self.conv2 = nn.Conv2d(in_channels=6, out_channels=6, kernel_size=3)
        self.fc1 = nn.Linear(6 * 6 * 6, 128)
        self.fc2 = nn.Linear(128, 10)

    def forward(self, x):
        x = self.conv1(x)
        x = F.relu(x)
        x = F.max_pool2d(x, 2)  # Correct pooling

        x = self.conv2(x)
        x = F.relu(x)
        x = F.max_pool2d(x, 2)  # Correct pooling

        x = x.view(x.size(0), -1)  # Flatten: batch_size x (flattened features)

        x = self.fc1(x)
        x = F.relu(x)

        x = self.fc2(x)
        return x

        
```

```python
model=cifar_cnn_pytorch()
criterion=nn.CrossEntropyLoss()
optimizer=optim.Adam(model.parameters(),lr=0.0001)
```

## TRAIN


```python
for epoch in range(10):
    for images,labels in train_loader:
        outputs=model(images)
        loss=criterion(outputs,labels)
        loss.backward()
        optimizer.step()
        optimizer.zero_grad()
    print(f"epoch:{epoch},loss:{loss.item()}")
```
